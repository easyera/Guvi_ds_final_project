```python
def summarize_categorical_feature(df, column):
    print(f"--- {column} ---")
    print(f"Unique values: {df[column].nunique()}")
    print(df[column].value_counts(dropna=False))
    print()

categorical_columns = ["status" ,"product", "category", "priority", "channel", "region", "subscription_type", "operating_system", "browser", "payment_method", "language", "customer_segment"]

for column in categorical_columns:
    summarize_categorical_feature(df, column)
```
