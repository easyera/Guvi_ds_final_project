```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/cleaned_data.csv')
```

```python
df.head()
```

```python
df['ticket_created_date'] = pd.to_datetime(df['ticket_created_date'])
```

```python
df['day_of_week'] = df['ticket_created_date'].dt.dayofweek
df['month'] = df['ticket_created_date'].dt.month
```

```python
cols_to_drop = [
    'ticket_created_date',
    'ticket_resolved_date',
    'resolution_notes',
    'issue_complexity_score',
    'status',
    'resolution_time_hours'
]
```

```python
df = df.drop(columns=cols_to_drop)
```

```python
df.columns
```

```python
df['ticket_rate'] = df['previous_tickets'] / df['customer_tenure_months']
```

```python
print(df['ticket_rate'].isin([float('inf')]).sum())
```

```python
df['ticket_rate'] = np.log1p(df['ticket_rate'])
```

```python
df.to_csv('../data/features/featured_data.csv', index=False)
```
